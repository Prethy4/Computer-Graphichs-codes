```python
x1 = int(input('x1:'))
y1 = int(input('y1:'))
x2 = int(input('x2:'))
y2 = int(input('y2:'))
```

```python
x_max = int(input()) 
y_max = int(input()) 
x_min = int(input()) 
y_min = int(input())
```

```python
m = ((y2-y1) / (x2-x1))
m
```

```python
intersection_points = []
if y_min <= y1 <= y_max:
    x = x1 + ((y_min - y1) / m )
    if x_min <= x <= x_max:
        intersection_points.append((x, y_min))
if y_min <= y2 <= y_max:
    x = x1 + ((y_max - y1) / m )
    if x_min <= x <= x_max:
        intersection_points.append((x, y_max))
if x_min <= x1 <= x_max:
    y = y1 + (m * (x_min - x1))
    if y_min <= y <= y_max:
        intersection_points.append((x_min, y))
if x_min <= x2 <= x_max:
    y = y1 + (m * (x_max - x1))
    if y_min <= y <= y_max:
        intersection_points.append((x_max, y))
print("Intersection points:", intersection_points)
```
