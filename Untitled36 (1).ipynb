```python
def inRegion(p, xmin, xmax, ymin, ymax):
  return xmin<=p[0]<=xmax and ymin<=p[1]<=ymax

window = [(0,3), (4,3), (4,0), (0, 0)]

xmin = window[0][0]
xmax = window[1][0]
ymin = window[3][1]
ymax = window[0][1]

print(f'xmin: {xmin},  xmax: {xmax},   ymin: {ymin},   ymax:{ymax}')

polygon = [(1,5), (4,5), (3,1), (2,1), (2,4), (1,1), (1,5)]

# for item in polygon:
#   print(item, end=' ')
#   print(inRegion(item, xmin, xmax, ymin, ymax))

subject_list = []

for i in range(len(polygon)-1):

  p1 = polygon[i]
  p2 = polygon[i+1]

  if (not inRegion(p1, xmin, xmax, ymin, ymax)) and (not inRegion(p2, xmin, xmax, ymin, ymax)) or (inRegion(p1, xmin, xmax, ymin, ymax) and inRegion(p2, xmin, xmax, ymin, ymax)):
    subject_list.append(p1)
  else:
    subject_list.append(p1)
    subject_list.append(['i',(i+i+1)/2])

print()
print('Subject_list:', subject_list)

print()
print('Clip_list:')
for i in range(len(subject_list)):
  if subject_list[i][0]=='i':
    print(subject_list[i], end=' ')
    i += 1
    while i<len(subject_list) and subject_list[i][0]!='i':
      print(subject_list[i], end=' ')
      i += 1
    if i<len(subject_list):
      print(subject_list[i])
    i += 1
```
