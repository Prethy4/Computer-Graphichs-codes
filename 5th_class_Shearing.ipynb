```python
import matplotlib.pyplot as plt
import matplotlib.image as img
import numpy as np
import cv2
import math
```

```python
img = cv2.imread("../thes.png",0)

w = img.shape[1] + 250
h = img.shape[0] + 250

shearX = float(input("Shear value: "))
shearY = 0

matrix = np.array([[1, shearX, 0], 
                   [0, 1, shearY]], dtype=np.float32)

sheared_image = cv2.warpAffine(img, matrix, (w, h))                 
                  
fig, axis = plt.subplots(1, 2, figsize=(7, 4))

axis[0].imshow(img)
axis[1].imshow(sheared_image)
axis[1].set_title('Sheared image')
  
plt.tight_layout()
plt.show()
```

```python
#translation

img = cv2.imread("../thes.png", 0)

w = img.shape[1] + 250
h = img.shape[0] + 250

tx = float(input("Translation in X: "))
ty = float(input("Translation in Y: "))

matrix = np.array([[1, 0, tx],
                   [0, 1, ty]], dtype=np.float32)

translated_image = cv2.warpAffine(img, matrix, (w, h))

fig, axis = plt.subplots(1, 2, figsize=(7, 4))

axis[0].imshow(img)
axis[1].imshow(translated_image)
axis[1].set_title('Translated image')

plt.tight_layout()
plt.show()
```

```python
#3D translation
n = int(input("Enter no. of co-ordinates:")) 

x = []
y = []
z = []

for i in range(n):
    ix = int(input("x value: "))
    iy = int(input("y value: "))
    iz = int(input("z value: "))
    
    x.append(ix)
    y.append(iy)
    z.append(iz)

tx = int(input("tx value: "))
ty = int(input("ty value: "))
tz = int(input("tz value: "))

for i in range(n):
    x[i] += tx
    y[i] += ty
    z[i] += tz

for i in range(n):
    print("(",x[i], ",", y[i],",", z[i], ")")
```

```python
# Rotation: x, y, and z-axis

n = int(input("Enter number of coordinates: "))

x = []
y = []
z = []
r = math.radians(float(input("Enter angle: ")))

for i in range(n):
    ix = float(input("Enter x: "))
    iy = float(input("Enter y: "))
    iz = float(input("Enter z: "))
    
    x.append(ix)
    y.append(iy)
    z.append(iz)

a = input("Select(x/y/z): ")

if a == 'x':
    for i in range(n):
        ny = y[i] * math.cos(r) - z[i] * math.sin(r)
        nz = y[i] * math.sin(r) + z[i] * math.cos(r)

        y[i] = ny
        z[i] = nz
    
elif a == 'y':
    for i in range(n):
        nx = x[i] * math.cos(r) + z[i] * math.sin(r)
        nz = -x[i] * math.sin(r) + z[i] * math.cos(r)

        x[i] = nx
        z[i] = nz
    
elif a == 'z':
    for i in range(n):
        nx = x[i] * math.cos(r) - y[i] * math.sin(r)
        ny = x[i] * math.sin(r) + y[i] * math.cos(r)

        x[i] = nx
        y[i] = ny

for i in range(n):
    print(f"New coordinates {i + 1}: ({x[i]}, {y[i]}, {z[i]})")
```

```python
# Scaling:

n = int(input("Enter number of coordinates: "))

x = []
y = []
z = []

for i in range(n):
    ix = int(input("Enter x: "))
    iy = int(input("Enter y: "))
    iz = int(input("Enter z: "))
    
    x.append(ix)
    y.append(iy)
    z.append(iz)

sx = int(input("Enter sx: "))
sy = int(input("Enter sy: "))
sz = int(input("Enter sz: "))

for i in range(n):
    x[i] *= sx
    y[i] *= sy
    z[i] *= sz

for i in range(n):
    print(f"New coordinates {i + 1}: ({x[i]}, {y[i]}, {z[i]})")
```

```python
# Shearing: x-axis, y-axi and z-axis

n = int(input("Enter number of coordinates: "))

x = []
y = []
z = []

for i in range(n):
    ix = int(input("Enter x: "))
    iy = int(input("Enter y: "))
    iz = int(input("Enter z: "))
    
    x.append(ix)
    y.append(iy)
    z.append(iz)
    
shx = float(input("Enter sx: "))
shy = float(input("Enter sy: "))
shz = float(input("Enter sz: "))

r = input("Select(x/y/z): ")

if r == 'x':
    for i in range(n):
        y[i] = y[i] + x[i] * shy
        z[i] = z[i] + x[i] * shz
elif r == 'y':
    for i in range(n):
        x[i] = x[i] + y[i] * shx
        z[i] = z[i] + y[i] * shz
elif r == 'z':
    for i in range(n):
        x[i] = x[i] + z[i] * shx
        y[i] = y[i] + z[i] * shy

for i in range(n):
    print(f"New coordinates {i + 1}: ({x[i]}, {y[i]}, {z[i]})")
```

```python
# Reflection: x-axis/yz plane, y-axis/xz plane, and z-axis/xy plane

n = int(input("Enter number of coordinates: "))

x = []
y = []
z = []

for i in range(n):
    ix = int(input("Enter x: "))
    iy = int(input("Enter y: "))
    iz = int(input("Enter z: "))
    
    x.append(ix)
    y.append(iy)
    z.append(iz)

r = input("Select(x/y/z): ")

if r == 'x':
    for i in range(n):
        x[i] = -x[i]
elif r == 'y':
    for i in range(n):
        y[i] = -y[i]
elif r == 'z':
    for i in range(n):
        z[i] = -z[i]

for i in range(n):
    print(f"New coordinates {i + 1}: ({x[i]}, {y[i]}, {z[i]})")
```
