```python
import numpy as np
from PIL import Image

input_image = np.array(Image.open('../thes.png'))

rotated_image1 = np.array(Image.fromarray(input_image).rotate(15, expand=True))
rotated_image2 = np.array(Image.fromarray(input_image).rotate(30, expand=True))
rotated_image3 = np.array(Image.fromarray(input_image).rotate(45, expand=True))
rotated_image4 = np.array(Image.fromarray(input_image).rotate(60, expand=True))
rotated_image5 = np.array(Image.fromarray(input_image).rotate(90, expand=True))


Image.fromarray(rotated_image1).save('rotate15.png')
Image.fromarray(rotated_image2).save('rotate30.png')
Image.fromarray(rotated_image3).save('rotate45.png')
Image.fromarray(rotated_image4).save('rotate60.png')
Image.fromarray(rotated_image5).save('rotate90.png')
```
