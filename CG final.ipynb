```python
#que 2
import math
n = int(input("Enter no. of co-ordinates: "))

x = []
y = []

r = math.radians(int(input("Enter theta: ")))

for i in range(n):
    xi = int(input("x: "))
    yi = int(input("y")) 
    
    x.append(xi)
    y.append(yi)
    
for i in range(n):
    x[i] = x[i]*math.cos(r) - y[i]*math.sin(r)
    y[i] = y[i]*math.cos(r) + x[i]*math.sin(r)
    
for i in range(n):
    print('(' + str(x[i]) +',' + str(y[i]) + ')')
```
