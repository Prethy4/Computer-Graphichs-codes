```python
#Digital Differential Algorithm
#input: (x1,y1) to (x2,y2):
x1 = int(input("x1: "))
y1 = int(input("y1: "))
x2 = int(input("x2: "))
y2 = int(input("y2: "))

dely = y2 - y1
delx = x2 - x1

m = dely / delx
print("m=",m)

upx = x1
upy = y1
if m>= 0 and m<= 1:
    upx = 1
    upy = m
elif m>= -1 and m<=0:
    upx = -1
    upy = -m
elif m>1:
    upx = 1/m
    upy = 1
elif m<-1:
    upx = -1/m
    upy = -1
    
print("upx=",upx)
print("upy=",upy)

x = x1
y = y1

while (1):
    print('(' + str(round(x)) + ',' + str(round(y)) + ')')
    x = x + upx
    y = y + upy
    
    if round(x)==x2 and round(y)==y2:
        print('(' + str(round(x)) + ',' + str(round(y)) + ')')
        break 
```

```python
#bresenhams algorithm
x1 = int(input("x1: "))
y1 = int(input("y1: "))
x2 = int(input("x2: "))
y2 = int(input("y2: "))

dely = y2 - y1
delx = x2 - x1

ipo = (2*dely) - delx
print("initial point=", ipo)

x = x1
y = y1

while(1):
    if ipo>=0:
        ipo = ipo + (2*dely) - (2*delx)
        x = x + 1 #round function not needed
        y = y + 1
        print('(' + str(round(x)) + ',' + str(round(y)) + ')')
    elif ipo<0:
        ipo = ipo + (2*dely)
        x = x + 1
        print('(' + str(round(x)) + ',' + str(round(y)) + ')')
    if round(x)==x2 and round(y)==y2:
        break
```

```python
#mid point circle
r = int(input("insert r: "))
x0 = 0
y0 = r 

ipo = 1 - r
print("initial point: ", ipo)

x = x0
y = y0

while(1):
    if ipo<0:
        x = x + 1
        ipo = ipo + (2*x) + 1
        print('(' + str(x) + ',' + str(y) + ')')
    elif ipo>=0:
        x = x + 1
        y = y - 1
        ipo = ipo + (2*x) + 1 - (2*y)
        print('(' + str(x) + ',' + str(y) + ')')
        
    if x >= y:
        break
```

```python
#3d rotation
import math
n = int(input('enter no of co-ordinates: '))

x = []
y = []
z = []
r = math.radians(int(input("Enter theta: ")))
a = input("type axis x/y/z: ")

for i in range(n):
    xi = int(input('x: '))    
    yi = int(input('y: '))    
    zi = int(input('z: '))
    
    x.append(xi)
    y.append(yi)
    z.append(zi)

if a == 'x':
    for i in range(n):
        y[i] = y[i]*math.cos(r) - z[i]*math.sin(r)
        z[i] = y[i]*math.sin(r) + z[i]*math.cos(r)
        
elif a == 'y':
    for i in range(n):
        x[i] = x[i]*math.cos(r) + z[i]*math.sin(r)
        z[i] = z[i]*math.cos(r) - x[i]*math.sin(r)
        
if a == 'z':
    for i in range(n):
        x[i] = x[i]*math.cos(r) - y[i]*math.sin(r)
        y[i] = x[i]*math.sin(r) + y[i]*math.cos(r)
        
for i in range(n):
    print('(' + str(x[i]) + ',' + str(y[i]) + ',' + str(z[i]) + ')')
```

```python
#3D shearing

import math
n = int(input('enter no of co-ordinates: '))

x = []
y = []
z = []

shx = int(input("Shx: "))
shy = int(input("Shy: "))
shz = int(input("Shz: "))

a = input("type axis x/y/z: ")

for i in range(n):
    xi = int(input('x: '))    
    yi = int(input('y: '))    
    zi = int(input('z: '))
    
    x.append(xi)
    y.append(yi)
    z.append(zi)

if a == 'x':
    for i in range(n):
        y[i] = y[i] + x[i]*shy
        z[i] = z[i] + x[i]*shz
        
elif a == 'y':
    for i in range(n):
        x[i] = x[i] + y[i]*shx
        z[i] = z[i] + y[i]*shz
        
if a == 'z':
    for i in range(n):
        x[i] = x[i] + z[i]*shx
        y[i] = x[i] + z[i]*shy
        
for i in range(n):
    print('(' + str(x[i]) + ',' + str(y[i]) + ',' + str(z[i]) + ')')
```

```python
#2D 30° rotation
import matplotlib.pyplot as plt
import matplotlib.image as img
import numpy as np
import cv2
import math

img = cv2.imread("../thes.png",0)
h, w = img.shape

theta = int(input('Enter theta: '))
nimg = np.zeros((w+280, h+280), dtype = "uint8")
radians = math.radians(theta)

for i in range(h):
    for j in range(w):
        pix = img[i][j]
        pnt = np.array([i,j]).reshape(-1,1)
        rotm = np.array([math.cos(radians), -math.sin(radians), math.sin(radians), math.cos(radians)])
        rotm = rotm.reshape(2,2)
        newmat = rotm @ pnt
        xf = newmat[0][0]
        yf = newmat[1][0]
        x2 = int(xf+350)
        y2 = int(yf)
        nimg[x2,y2] = pix

print(nimg.shape)
fig, ax = plt.subplots()
ax.imshow(nimg, extent = [0, 400, 250, -100], cmap = 'gray')
plt.title("30° rotation")
plt.show()
```

```python
#Colen Sutherland Line Clipping Algorithm
x1 = int(input('x1:'))
y1 = int(input('y1:'))
x2 = int(input('x2:'))
y2 = int(input('y2:'))

x_max = int(input("x_max: ")) 
y_max = int(input("y_max: ")) 
x_min = int(input("x_min: ")) 
y_min = int(input("y_min: "))

m = ((y2-y1) / (x2-x1))
m

intersection_points = []
if y_min <= y1 <= y_max:
    x = x1 + ((y_min - y1) / m )
    if x_min <= x <= x_max:
        intersection_points.append((x, y_min))
if y_min <= y2 <= y_max:
    x = x1 + ((y_max - y1) / m )
    if x_min <= x <= x_max:
        intersection_points.append((x, y_max))
if x_min <= x1 <= x_max:
    y = y1 + (m * (x_min - x1))
    if y_min <= y <= y_max:
        intersection_points.append((x_min, y))
if x_min <= x2 <= x_max:
    y = y1 + (m * (x_max - x1))
    if y_min <= y <= y_max:
        intersection_points.append((x_max, y))
print("Intersection points:", intersection_points)

if x > x_max and y > y_max:
    top = 1
    bottom = 0
    right = 1 
    left = 0
elif x > x_max and y < y_min:
    top = 0
    bottom = 1
    right = 1 
    left = 0
elif x > x_max and y_min < y < y_max:
    top = 0
    bottom = 0
    right = 1 
    left = 0
elif x_min < x < x_max and y > y_max:
    top = 1
    bottom = 0
    right = 0
    left = 0
elif x_min < x < x_max and y < y_min:
    top = 0
    bottom = 1
    right = 0
    left = 0
elif x_min < x < x_max and y_min < y < y_max:
    top = 0
    bottom = 0
    right = 0
    left = 0
elif x < x_min and y > y_max:
    top = 1
    bottom = 0
    right = 0 
    left = 1
elif x < x_min and y < y_min:
    top = 0
    bottom = 1
    right = 0 
    left = 1
elif x < x_min and y_min < y < y_max:
    top = 0
    bottom = 0
    right = 1 
    left = 0
    
print(top, bottom, right, left)
```

```python
#weiler atherton
window = [(0,3),(4,3),(4,0),(0,0)]

x_max = 4
y_max = 3
x_min = 0
y_min = 0

polygon = [(1,5),(4,5),(3,1),(2,1),(2,4),(1,1), (1,5)]

subject_list = []

for item in range(len(polygon) - 1):
    x = polygon[item]
    y = polygon[item+1]

    in_x = x_min <= x[0] <= x_max and y_min <= x[1] <= y_max
    in_y = x_min <= y[0] <= x_max and y_min <= y[1] <= y_max

    if (not in_x and not in_y) or (in_x and in_y):
        subject_list.append(x)
    else:
        subject_list.append(x)
        subject_list.append('i')
print("subject list", subject_list)

clip_list = []

for item in range(len(window) - 1):
    x = window[item]
    y = window[item+1]

    if (x and y) and (x not in clip_list and y not in clip_list):
        clip_list.append(x)
        for i in range(window[0][0], window[1][0]) or range(window[0][1], window[1][1]):
            clip_list.append(f"i{i}")
    clip_list.append(y)

print("clip list: ", clip_list)
```
