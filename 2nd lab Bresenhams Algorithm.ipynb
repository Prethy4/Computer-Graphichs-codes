```python
x1 = int(x1)
y1 = int(y1)
x2 = int(x2)
y2 = int(y2)
```

```python
delx = x2 - x1
dely = y2 - y1
ipo = (2*dely - delx)
print(ipo)
```

```python
x = x1 
y = y1
while(1):
    if ipo>=0:
        ipo = ipo + (2*dely) - (2*delx)
        x = x + 1
        y = y + 1
        print("(" + str(x) + "," + str(y) + ")")
    elif ipo<0:
        ipo = ipo+ (2*dely)
        x = x + 1
        print("(" + str(x) + "," + str(y) + ")")
    if x ==x2 and y == y2:
        break
        
```
