```python
## Cuhen Sutherland Line clipping algorithm

x1, y1 = input().split()
x2, y2 = input().split()


x1 = int(x1)
x2 = int(x2)
y1 = int(y1)
y2 = int(y2)

m = (y2-y1)/(x2-x1)
print(m)

Xmin = int(input("Xmin: "))
Xmax = int(input("Xmax: "))
Ymin = int(input("Ymin: "))
Ymax = int(input("Ymax: "))


B = x1 + ((Ymin - y1)/m)
if Xmin <= B <= Xmax:
    print(B, Ymin)
    
B = x1 + ((Ymax - y1)/m)
if Xmin <= B <= Xmax:
    print(B, Ymax)

A = y1 + (m*(Xmin - x1))
if Ymin <= A <= Ymax:
    print(Xmin, A)
    
A = y1 + (m*(Xmax - x1))
if Ymin <= A <= Ymax:
    print(Xmax, A)   


def findRegion(x, y):
    ans = [0]*4
    
    if x < Xmin and y > Ymax:
        ans = [1,0,0,1]
        
    elif Xmin < x < Xmax and y > Ymax:
        ans = [1,0,0,0]
        
    elif x > Xmax and y > Ymax:
        ans = [1,0,1,0]
        
    elif x < Xmin and Ymin < y < Ymax:
        ans = [0,0,0,1]
        
    elif Xmin < x < Xmax and Ymin < y< Ymax:
        ans = [0,0,0,0]
        
    elif x > Xmax and Ymin < y < Ymax:
        ans = [0,0,1,0]
        
    elif x < Xmin and y < Ymin:
        ans = [0,1,0,1]
        
    elif Xmin < x < Xmax and y < Ymin:
        ans = [0,1,0,0]
        
    elif x > Xmax and y < Ymin:
        ans = [0,1,1,0]
        
    return ans

r1 = findRegion(x1, y1)
r2 = findRegion(x2, y2)

print(r1, r2)

#perform OR operation on r1, r2. if ans is 0000, line visible
ans = []
for i in range(4):
    ans.append(r1[i] or r2[i])
    
print(ans)
if ''.join([str(x) for x in ans])=='0000':
    print('visible')
    
else:
    #perform AND operation on r1, r2.
    ans = []
    for i in range(4):
        ans.append(r1[i] and r2[i])
        
    if ''.join([str(x) for x in ans])!='0000':
        print('invisible')
        
    else:
        print('clipped')
        print(ans)
```
